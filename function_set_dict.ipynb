```python
a = {1,2,3,4,5}
b= {3,4,5,6,7,8 , 9}


d = a.union(b)
print("union: " , d)
c = a.intersection(b)
print("intersection : " , c)
```

```python
# ALL FUNCTION OF SET ;

# Creation ----
# set()
# set(iterable)

# Add / Update -----
# add(x)
# update(iterable)

# Remove //
# remove(x)
# discard(x)
# pop()
# clear()

# Copy.  -------
# copy()

# Set Operations----------
# union(other)                    # |
# intersection(other)             # &
# difference(other)               # -
# symmetric_difference(other)     # ^

# Update Set Operations---------
# intersection_update(other)
# difference_update(other)
# symmetric_difference_update(other)

# Relations------------
# issubset(other)                 # <=
# issuperset(other)               # >=
# isdisjoint(other)

# Built-in Functions commonly used with sets---------
# len(s)
# min(s)
# max(s)
# sum(s)
# sorted(s)

# Membership-----------
# x in s
# x not in s
```

```python
#  FUNCTION OF DICTIONARY


# # Creation
# dict()

# # Add / Update
# d[key] = value
# update(other_dict)

# # Access
# get(key)
# get(key, default)

# # Remove
# pop(key)
# pop(key, default)
# popitem()
# clear()
# del d[key]

# # Copy
# copy()

# # Views
# keys()
# values()
# items()

# # Utility
# setdefault(key)
# setdefault(key, default)

# # Create from iterable
# fromkeys(iterable)
# fromkeys(iterable, value)

# # Built-in Functions commonly used with dictionaries
# len(d)
# min(d)
# max(d)
# sorted(d)
# sum(d)          # works on numeric keys

# # Membership
# key in d
# key not in d

# # Iteration
# for key in d:
#     pass

# for value in d.values():
#     pass

# for key, value in d.items():
#     pass
```
