```python
# to find the number of applications we know that all loan will have unique id 
display(df['id'].nunique())
```

```python
app_in_current_month= df[df['issue_date'].dt.month==6]
# ye hame june ke moth me kitne applications aaye hai wo de dega so if u want to find total no of applications in the month u are in than 
# current_date = pd.Timestamp.today()

#current_month = current_date.month
#app_in_current_month= df[df['issue_date'].dt.month==current_month]
num_appliactions_in_current_month=app_in_current_month['id'].nunique()
print(num_appliactions_in_current_month)
```

```python
#for calculating loans sanctioned till date 
curr_date = pd.Timestamp.today()
curr_month = curr_date.month
curr_day = curr_date.day

curr_month = df[(df['issue_date'].dt.month == curr_month) & (df['issue_date'].dt.day <= curr_day)]
curr_month = curr_month['id'].nunique()
print(curr_month)
```

```python
curr_date = pd.Timestamp.today()
current_month_number = curr_date.month
df_current_month_loans = df[df['issue_date'].dt.month == current_month_number]
total_loan_amount_current_month = df_current_month_loans['loan_amount'].sum()
total_loan_amount_current_month_in_millions = total_loan_amount_current_month/1000000
print(f"Total loan amount for current month: {total_loan_amount_current_month_in_millions:,.2f}M$")
```

```python
loan_status_data = df.groupby(['loan_status']).agg({
    'id':'count',
    'dti':'mean',
    'loan_amount':'sum',
    'int_rate':'mean',
    'total_payment':'sum'
})


loan_status_data.rename(columns={'id':'loan_count'},inplace=True)


print(loan_status_data)
```

```python
loan_status_data = df.groupby(['loan_status']).agg({
    'id':'count',
    'dti':'mean',
    'loan_amount':'sum',
    'int_rate':'mean',
    'total_payment':'sum'
})

loan_status_data.rename(columns={'id':'loan_count'},inplace=True)

no_of_app = loan_status_data.loc['Current','loan_count']+loan_status_data.loc['Fully Paid','loan_count']

print(f"Number of unique IDs for 'Current' or 'Fully Paid' loans: {no_of_app}")
```
