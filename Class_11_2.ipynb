```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns


sns.set_theme(style="whitegrid")


x = [1, 2, 3, 4, 5]
y = [10, 15, 13, 20, 25]


sns.lineplot(x=x, y=y, marker='o', color='purple', linewidth=2)

plt.title("Line Chart")
plt.xlabel("X Axis")
plt.ylabel("Y Axis")
plt.tight_layout()
plt.show()
```

```python
subjects = ["hindi", "english", "maths", "programming"]
scores = [88, 97, 95, 92]


sns.barplot(x=subjects, y=scores, color='grey')

plt.title("Some random bs")
plt.xlabel("rbs1")
plt.ylabel("rbs2")

plt.tight_layout()
plt.show()
```

```python

data = np.random.randn(500)


sns.histplot(data, bins=20, kde=True, color='teal')

plt.title("Histogram with Density Curve")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()
```

```python

x = np.random.rand(50)
y = x + np.random.normal(0, 0.1, 50)

sns.scatterplot(x=x, y=y, color='coral', s=100, edgecolor='black')

plt.title("Scatter Plot")
plt.xlabel("Variable X")
plt.ylabel("Variable Y")
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns


sns.set_theme(style="whitegrid")


batch1 = np.random.randint(10, 70, 100)

sns.boxplot(y=batch1, color="skyblue")

plt.tight_layout()
plt.show()
```
